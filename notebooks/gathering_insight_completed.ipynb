What were the most popular female and male baby names in California in each year ?

```python
import pandas as pd
```

```python
state_babies = pd.read_csv('state_baby_names.csv')
```

```python
state_babies.head()
```

```python
ca_babies = state_babies.loc[state_babies['State'] == 'CA' , :]
```

```python
def popular(s): 
    '''Recieves s, a Panda Series, containing baby names in order of highest count to lowest count.
    Returns the most popular name in s.'''
    return s.iloc[0]
```

```python
#Grouping CA babies by both year and gender
ca_grouped = ca_babies.sort_values('Count', ascending = False).groupby(['Year','Gender']).agg(popular)
```
